```
CREATE AUTHENTICATION POLICY AP_PAT_USER
  AUTHENTICATION_METHODS = ('PROGRAMMATIC_ACCESS_TOKEN');
```

```
CREATE OR REPLACE NETWORK POLICY NP_ALLOWED_IPS
  ALLOWED_IP_LIST = ('255.255.255.255/XX');
```

```
//Generate token for user with SYSADMIN access and 15 days of expiry
ALTER USER PAT_DEMO_USER
    ADD PROGRAMMATIC ACCESS TOKEN PAT_TOKEN
    ROLE_RESTRICTION = 'ROLE_STREAMLIT_APP'
    DAYS_TO_EXPIRY = 15
    COMMENT = 'Token Generation for User used in Automation'
```
